```python
df2["type"].value_counts()
```

```python
# Replace AUTOMOBILE by 1 and TRUCK by 0
df2["type_encoded"] = df2["type"].replace(to_replace=['Automobile','Truck'],value=[1,0])
df2
```

```python
# MANUFACT :
```

```python
percentage_freq = df2["manufact"].value_counts()*100/len(df2)
```

```python
df2["manufact"].nunique()
```

```python
percentage_freq
```

```python
# lets create a category to classify all manufact values having less than 2% of percentage frequence
df2["manufact_rare_wt"] = df2["manufact"].replace(to_replace=['Subaru','Jaguar','Saab','Ininiti'],value='others')
```

```python
df2['manufact_rare_wt'].value_counts()
```

```python
manufact_mean = df2.groupby("manufact_rare_wt")["price"].mean()
```

```python
manufact_mean
```

```python
# Need to MAP the manufact_mean to each category
df2["manufact_en"]=df2["manufact_rare_wt"].map(manufact_mean)
```
